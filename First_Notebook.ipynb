```python
# ask what time it is
hours = int(input("how many hours after midnight is it? "))
mins = int(input("how many minutes into this hour is it? "))
# ask how many hours into the future user would like to go
futhours = int(input("hOW many hours into future would you like to go? "))
futmins = int(input("hOW many minutes into future would you like to go? "))
# add hours to what time it is
totalhours = hours + futhours
totalmins = mins + futmins
#find remainder when divided by 24
hoursrem = totalhours%24
minsrem = totalmins%60
#print final time
print("The time", futhours,"hours and", futmins, "minutes into the future will be",hoursrem, ":", minsrem)
```
